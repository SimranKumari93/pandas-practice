```python
import pandas as pd
```

```python
df = pd.read_csv('search_logs.csv')
```

```python
df.columns
```

```python
df.shape
```

```python
df.replace({
    'False': 0, 'True': 1, 'yes': 1, 'No': 0
})
```

```python
for col in df.columns:
    if df[col].dtype == 'object':
        df[col].fillna(df[col].mode()[0], inplace=True)
    else:
        df[col].fillna(df[col].median(), inplace=True)
```

```python
df.dropna(thresh=len(df.columns) - 3, inplace=True)
```

```python
df.replace(r'^\s*$', pd.NA, regex=True, inplace=True)
```

```python
df.replace({True: 1, False: 0, 'True': 1, 'False': 0}, inplace=True)
```

```python
df.to_csv("search_cleaned.csv", index=False)
```
